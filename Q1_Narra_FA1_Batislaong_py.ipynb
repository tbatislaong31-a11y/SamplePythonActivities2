```python
import math

#Ask the user to input the variable values
x1 = float(input("enter x1 please: "))
x2 = float(input("enter x2 please: "))
y1 = float(input("enter y1 please: "))
y2 = float(input("enter y2 please: "))

#calculatethe distance usingthe given distance formula
distance = math.sqrt(math.pow(x2-x1, 2)) + math.pow(y2-y1, 2)
print(f"the distance is: {distance}")

#1. the math library helped simplify my code through applying operations on all the numbers rather than applying to each and every one of them which takes more time and makes it more complex.
#2. The functions that were easy to use are math.pow and math.sqrt because it makes square root and power easier to apply, yknow? beacuse iamgine if you had to manually apply each and every ooperation.
#3. The progran would be so umch complex, longer, and harder to understand without pow() and sqrt() because you would have to apply each and every operation such as square root to each and every one of the digits inside of the code
```
